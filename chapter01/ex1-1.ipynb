```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
# Download and prepare the data
data_root = "https://github.com/ageron/data/raw/main/"
lifesat = pd.read_csv(data_root + "lifesat/lifesat.csv")
# inspect columns to confirm exact names
print('lifesat columns ->', lifesat.columns.tolist())
# The CSV uses the column name 'GDP per capita (USD)' for GDP per capita
X = lifesat[["GDP per capita (USD)"]].values
y = lifesat["Life satisfaction"].values
# Debug prints: shapes and a small sample
print('X shape:', X.shape)
print('y shape:', y.shape)
# If a Country column exists, print labeled rows (Country, GDP, Life satisfaction)
if 'Country' in lifesat.columns:
    print('First 5 rows with Country, GDP per capita (USD), Life satisfaction:')
    print(lifesat.loc[:4, ['Country', 'GDP per capita (USD)', 'Life satisfaction']])
else:
    print('X first 5 rows:\n', X[:5])
    print('y first 5 values:\n', y[:5])
```

```python
# Visualize the data
lifesat.plot(kind="scatter", grid=True, x="GDP per capita (USD)", y="Life satisfaction")
plt.axis([23_500, 62_500, 4, 9])
plt.xlabel("GDP per capita")
plt.ylabel("Life satisfaction")
plt.show()
```

```python
# Select and train a model
model = LinearRegression()
model.fit(X, y)
```
